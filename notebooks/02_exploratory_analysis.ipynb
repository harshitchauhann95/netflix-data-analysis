# 📊 02. Netflix Exploratory Data Analysis (EDA)

In-depth analysis of catalog composition, release trends, geographic production, and ratings.

```
import pandas as pd
import numpy as np
import os
import sys

sys.path.append(os.path.abspath('..'))
from src.data_loader import load_processed_data
```

```
# Load processed data
df = load_processed_data('../data/processed/netflix_cleaned.csv')
df.head(3)
```

```
# 1. Movies vs TV Shows breakdown
type_summary = df['type'].value_counts()
type_pct = df['type'].value_counts(normalize=True) * 100
pd.DataFrame({'Count': type_summary, 'Share (%)': type_pct.round(2)})
```

```
# 2. Top Content Producing Countries
countries = df['country'].dropna().apply(lambda x: [c.strip() for c in str(x).split(',') if c.strip() != 'Unknown'])
flat_countries = pd.Series([c for sub in countries for c in sub])
top_10_countries = flat_countries.value_counts().head(10)
print('--- Top 10 Countries ---')
print(top_10_countries)
```

```
# 3. Release Year Trends
df[df['release_year'] >= 2010].groupby(['release_year', 'type']).size().unstack(fill_value=0)
```
