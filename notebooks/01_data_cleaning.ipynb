# 🧹 01. Netflix Data Cleaning & Preprocessing

This notebook handles missing data imputation, date parsing, and feature engineering for the Netflix content catalog.

```
import pandas as pd
import numpy as np
import os
import sys

# Ensure src modules are accessible
sys.path.append(os.path.abspath('..'))
from src.data_loader import load_raw_data
from src.cleaning import clean_netflix_data
```

```
# Load raw dataset
df_raw = load_raw_data('../data/raw/netflix_titles.csv')
print(f'Initial Shape: {df_raw.shape}')
df_raw.head(3)
```

```
# Check missing value statistics
missing = df_raw.isnull().sum()
pct_missing = (missing / len(df_raw)) * 100
pd.DataFrame({'Missing Count': missing, 'Percentage (%)': pct_missing.round(2)})
```

```
# Clean dataset using modular cleaning routine
processed_path = '../data/processed/netflix_cleaned.csv'
df_cleaned = clean_netflix_data(df_raw, save_path=processed_path)
print(f'Cleaned Dataset Shape: {df_cleaned.shape}')
df_cleaned.head(3)
```
